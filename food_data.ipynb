```python
import pandas as pd
import json
import os

try:
    df = pd.read_csv('식품_일반음식점_충청남도.csv', encoding='cp949', low_memory=False)
except:
    df = pd.read_csv('식품_일반음식점_충청남도.csv', encoding='euc-kr', low_memory=False)

clean_df = pd.DataFrame({
    '사업장명': df.iloc[:, 8],
    '업태구분명': df.iloc[:, 9],
    '도로명주소': df.iloc[:, 19]
}).dropna()

# HTML 찌꺼기 완벽 차단
clean_df = clean_df[~clean_df['사업장명'].astype(str).str.contains('<|>|meta|html|A premium', case=False, na=False)]

records = clean_df.to_dict(orient='records')

# src/data 폴더에 .ts 파일로 저장
output_dir = 'src/data'
os.makedirs(output_dir, exist_ok=True)
file_path = os.path.join(output_dir, 'food_data.ts')

ts_content = f"export const FOOD_DATA = {json.dumps(records, ensure_ascii=False, indent=2)};\n"

with open(file_path, 'w', encoding='utf-8-sig') as f:
    f.write(ts_content)

print(f"파일 생성 완료: {file_path} (총 {len(records)}개)")
```
